## 1️⃣ One-Hot Encoding

### 📖 Recap
**One-Hot Encoding** converts a categorical column with `N` unique categories into `N` new binary (0/1) columns.

| Color  | → | Color_Red | Color_Blue | Color_Green |
|--------|---|-----------|------------|-------------|
| Red    |   | 1         | 0          | 0           |
| Blue   |   | 0         | 1          | 0           |
| Green  |   | 0         | 0          | 1           |

**When to use:**
- Nominal (unordered) categories
- Low cardinality (few unique values)
- Algorithms sensitive to magnitude (KNN, SVM, Linear Regression)

**Python tool:** `pd.get_dummies(df, columns=['col'])`

```python
# 💡 WORKED EXAMPLE — One-Hot Encoding

df_example = pd.DataFrame({
    'Name':  ['Alice', 'Bob', 'Carol', 'Dave'],
    'Color': ['Red', 'Blue', 'Green', 'Red'],
    'Score': [85, 90, 78, 92]
})

print('BEFORE encoding:')
print(df_example)
print()

df_ohe = pd.get_dummies(df_example, columns=['Color'])

print('AFTER One-Hot Encoding:')
print(df_ohe)
```

## 2️⃣ Dummy Encoding

### 📖 Recap
**Dummy Encoding** is One-Hot Encoding minus one column (`drop_first=True`). This avoids the **dummy variable trap** (multicollinearity) — where one column is perfectly predictable from the others.

| Color  | → | Color_Blue | Color_Green | *(Red implied when both = 0)* |
|--------|---|------------|-------------|------------------------------|
| Red    |   | 0          | 0           |                              |
| Blue   |   | 1          | 0           |                              |
| Green  |   | 0          | 1           |                              |

**When to use:**
- Linear Regression, Logistic Regression (multicollinearity matters)
- Any model that can be affected by correlated features

**Python tool:** `pd.get_dummies(df, columns=['col'], drop_first=True)`

```python
# 💡 WORKED EXAMPLE — Dummy Encoding

df_dummy = pd.DataFrame({
    'Name':  ['Alice', 'Bob', 'Carol', 'Dave'],
    'Color': ['Red', 'Blue', 'Green', 'Red']
})

print('One-Hot (3 columns — has multicollinearity):')
print(pd.get_dummies(df_dummy, columns=['Color']))
print()

print('Dummy Encoding (2 columns — no multicollinearity):')
print(pd.get_dummies(df_dummy, columns=['Color'], drop_first=True))
print()
print("→ 'Red' is the reference: when Color_Blue=0 AND Color_Green=0, it's Red")
```

## 3️⃣ Label Encoding

### 📖 Recap
**Label Encoding** assigns a unique integer to each category **in-place** (no new columns created).

| Education   | → | Encoded |
|-------------|---|---------|
| High School |   | 0       |
| Bachelor    |   | 1       |
| Master      |   | 2       |
| PhD         |   | 3       |

**When to use:**
- ✅ **Ordinal** categories (natural order: Low < Medium < High)
- ✅ Target/label column (always label-encode the y variable)
- ❌ Do NOT use for nominal categories (City, Country) — the model thinks Paris > London!

**Python tool:** `sklearn.preprocessing.LabelEncoder`

```python
# 💡 WORKED EXAMPLE — Label Encoding

df_label = pd.DataFrame({
    'Student':   ['Alice', 'Bob', 'Carol', 'Dave', 'Eve'],
    'Education': ['Bachelor', 'PhD', 'Master', 'High School', 'Bachelor'],
    'Score':     [78, 92, 85, 65, 80]
})

print('BEFORE:')
print(df_label)
print()

le = LabelEncoder()
# Fit on ordered categories
le.fit(['High School', 'Bachelor', 'Master', 'PhD'])
df_label['Education_encoded'] = le.transform(df_label['Education'])

print('AFTER Label Encoding:')
print(df_label)
print()
print('Mapping:', dict(zip(le.classes_, le.transform(le.classes_))))
```

### 🏋️ Exercise 3 — Label Encoding

You have a product quality dataset with an ordinal column `'Quality'`.

**Tasks:**
1. Apply Label Encoding to the `'Quality'` column.
2. Make sure the encoding respects the natural order: `Low=0, Medium=1, High=2, Premium=3`.
3. Print before and after.
4. Why would using One-Hot Encoding here be a mistake?

```python
# 🏋️ EXERCISE 3

products = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Tablet', 'Watch', 'Camera', 'Speaker'],
    'Quality': ['High', 'Medium', 'Low', 'Premium', 'High', 'Low'],
    'Price':   [1200, 800, 400, 300, 950, 150]
})

print('BEFORE:')
print(products)
print()

# Task 1 & 2: Apply Label Encoding respecting natural order
# Order: Low=0, Medium=1, High=2, Premium=3
#
# NOTE: sklearn's LabelEncoder ignores the order you pass to .fit() — it sorts
# alphabetically. To enforce a specific ordinal order we use a manual mapping.
order_map = {'Low': 0, 'Medium': 1, 'High': 2, 'Premium': 3}
products['Quality_encoded'] = products['Quality'].map(order_map)

# Task 3: Print before and after
print('AFTER Label Encoding:')
print(products)
print()
print('Mapping:', order_map)

# Task 4: Why would One-Hot Encoding be a mistake here?
# Answer: Quality is ordinal (Low < Medium < High < Premium). One-Hot Encoding
# would throw away that ordering by treating each level as an unrelated category,
# so the model would lose useful information about the rank between quality tiers.
```

## 4️⃣ Standard Scaling (Z-score Normalization)

### 📖 Recap
**Standard Scaling** transforms features to have **mean = 0** and **standard deviation = 1**.

$$z = \frac{x - \mu}{\sigma}$$

| Salary   | → | Scaled  |
|----------|---|---------|
| 30,000   |   | -1.41   |
| 50,000   |   | -0.71   |
| 70,000   |   |  0.00   |
| 90,000   |   |  0.71   |
| 110,000  |   |  1.41   |

**When to use:**
- ✅ Normal (Gaussian) distribution
- ✅ PCA, SVM, Logistic Regression, Linear Regression
- ✅ Less sensitive to outliers than Min-Max

**Python tool:** `sklearn.preprocessing.StandardScaler`

```python
# 💡 WORKED EXAMPLE — Standard Scaling

df_scale = pd.DataFrame({
    'Employee': ['A', 'B', 'C', 'D', 'E'],
    'Salary':   [30000, 50000, 70000, 90000, 110000],
    'Age':      [25, 30, 35, 40, 45]
})

print('BEFORE scaling:')
print(df_scale)
print()

scaler = StandardScaler()
df_scale[['Salary_scaled', 'Age_scaled']] = scaler.fit_transform(df_scale[['Salary', 'Age']])

print('AFTER Standard Scaling:')
print(df_scale.round(3))
print()
print(f'Salary mean: {df_scale["Salary_scaled"].mean():.6f}  (should be ≈ 0)')
print(f'Salary std:  {df_scale["Salary_scaled"].std():.6f}   (should be ≈ 1)')
```

### 🏋️ Exercise 4 — Standard Scaling

You have a medical dataset. Apply **Standard Scaling** to the numeric columns.

**Tasks:**
1. Apply Standard Scaling to `'Height'`, `'Weight'`, and `'BloodPressure'` columns.
2. Print before and after.
3. Verify: Print the mean and std of each scaled column (should be ≈ 0 and ≈ 1).
4. Identify which original feature had the largest variance. Does scaling affect this?

```python
# 🏋️ EXERCISE 4

medical = pd.DataFrame({
    'PatientID':     [1, 2, 3, 4, 5, 6],
    'Height':        [165, 172, 158, 180, 175, 168],   # cm
    'Weight':        [55, 80, 62, 95, 70, 63],          # kg
    'BloodPressure': [120, 145, 110, 160, 130, 125],    # mmHg
    'Diabetic':      [0, 1, 0, 1, 0, 0]
})

print('BEFORE scaling:')
print(medical)
print()

# Task 1: Apply Standard Scaling to Height, Weight, BloodPressure
features_to_scale = ['Height', 'Weight', 'BloodPressure']

scaler = StandardScaler()
medical[['Height_scaled', 'Weight_scaled', 'BloodPressure_scaled']] = scaler.fit_transform(
    medical[features_to_scale]
)

# Task 2: Print before and after
print('AFTER Standard Scaling:')
print(medical.round(3))
print()

# Task 3: Verify mean ≈ 0, std ≈ 1 for each scaled column
for col in ['Height_scaled', 'Weight_scaled', 'BloodPressure_scaled']:
    print(f'{col}: mean={medical[col].mean():.6f}, std={medical[col].std():.6f}')

# Task 4: Which feature had the largest variance before scaling?
print()
print('Variance BEFORE scaling:')
print(medical[features_to_scale].var().round(2))
# Answer: BloodPressure has the largest variance (range 110–160).
# Scaling removes these magnitude differences — every scaled column ends up with
# the same variance, so no single feature dominates distance-based or
# gradient-based models just because its raw values are bigger.
```

## 5️⃣ Min-Max Scaling (Normalization)

### 📖 Recap
**Min-Max Scaling** transforms features to a **fixed range [0, 1]**.

$$x' = \frac{x - x_{\min}}{x_{\max} - x_{\min}}$$

| Temperature | → | Scaled |
|-------------|---|--------|
| 5           |   | 0.00   |
| 10          |   | 0.10   |
| 25          |   | 0.40   |
| 40          |   | 0.70   |
| 55          |   | 1.00   |

**When to use:**
- ✅ Neural networks (inputs should be in [0, 1])
- ✅ Image pixel values (0–255 → 0.0–1.0)
- ✅ When you need bounded outputs
- ⚠️ Sensitive to outliers — one extreme value squashes all others!

**Python tool:** `sklearn.preprocessing.MinMaxScaler`

```python
# 💡 WORKED EXAMPLE — Min-Max Scaling

df_mm = pd.DataFrame({
    'City':        ['Tashkent', 'Moscow', 'Berlin', 'Tokyo', 'London'],
    'Temperature': [5, 10, 25, 40, 55]   # °C
})

print('BEFORE Min-Max Scaling:')
print(df_mm)
print()

mm_scaler = MinMaxScaler()
df_mm['Temp_scaled'] = mm_scaler.fit_transform(df_mm[['Temperature']])

print('AFTER Min-Max Scaling:')
print(df_mm)
print()
print(f'Min: {df_mm["Temp_scaled"].min():.2f}  (should be 0.00)')
print(f'Max: {df_mm["Temp_scaled"].max():.2f}  (should be 1.00)')
```

### 🏋️ Exercise 5 — Min-Max Scaling

You have a housing dataset. Apply **Min-Max Scaling** to numeric features.

**Tasks:**
1. Apply Min-Max Scaling to `'Size'`, `'Rooms'`, and `'Price'`.
2. Print before and after.
3. Verify: min = 0.0, max = 1.0 for each scaled column.
4. **Bonus:** Add an outlier row (`Price = 5,000,000`) and re-scale. What happens to other houses?

```python
# 🏋️ EXERCISE 5

housing = pd.DataFrame({
    'HouseID': [1, 2, 3, 4, 5],
    'Size':    [60, 90, 120, 75, 200],     # m²
    'Rooms':   [2, 3, 4, 2, 5],
    'Price':   [80000, 150000, 220000, 110000, 350000]  # USD
})

features = ['Size', 'Rooms', 'Price']

# Task 2 (print BEFORE)
print('BEFORE Min-Max Scaling:')
print(housing)
print()

# Task 1: Apply Min-Max Scaling to Size, Rooms, AND Price
mm_scaler = MinMaxScaler()
housing[['Size_scaled', 'Rooms_scaled', 'Price_scaled']] = mm_scaler.fit_transform(
    housing[features]
)

# Task 2 (print AFTER)
print('AFTER Min-Max Scaling:')
print(housing.round(3))
print()

# Task 3: Verify min=0.0, max=1.0 for each scaled column
for col in ['Size_scaled', 'Rooms_scaled', 'Price_scaled']:
    print(f'{col}: min={housing[col].min():.2f}, max={housing[col].max():.2f}')

# BONUS Task 4: Add an outlier and observe the effect
print()
print('--- Bonus: adding an outlier house (Price = 5,000,000) ---')

housing_outlier = pd.concat(
    [housing[['HouseID', 'Size', 'Rooms', 'Price']],
     pd.DataFrame([{'HouseID': 6, 'Size': 100, 'Rooms': 3, 'Price': 5_000_000}])],
    ignore_index=True
)
housing_outlier['Price_scaled'] = MinMaxScaler().fit_transform(housing_outlier[['Price']])

print(housing_outlier.round(4))
print()
print('Observation: the five original houses (which span 80k–350k) are now squashed')
print('into the range ~0.00–0.05, because one extreme value (5,000,000) stretches')
print('the [min, max] range. Min-Max Scaling is very sensitive to outliers — for')
print('skewed data with extremes, StandardScaler or RobustScaler is safer.')
```

## 🏆 FINAL CHALLENGE — Full Pipeline

Build a complete feature engineering pipeline combining encoding AND scaling on a real-world dataset.

**Dataset:** Employee attrition (will someone quit?)

**Tasks:**
1. Apply **Label Encoding** to `'Satisfaction'` (ordinal: Low < Medium < High)
2. Apply **Dummy Encoding** to `'Department'` (nominal)
3. Apply **Standard Scaling** to `'Salary'` and `'YearsAtCompany'`
4. Print the final processed DataFrame
5. **Think:** Which encoding/scaling choice would change if we used a Random Forest instead of Logistic Regression?

```python
# 🏆 FINAL CHALLENGE

employees = pd.DataFrame({
    'EmployeeID':      [1, 2, 3, 4, 5, 6, 7, 8],
    'Department':      ['IT', 'HR', 'Finance', 'IT', 'HR', 'Finance', 'IT', 'HR'],
    'Satisfaction':    ['High', 'Low', 'Medium', 'High', 'Low', 'High', 'Medium', 'Low'],
    'Salary':          [75000, 45000, 60000, 90000, 40000, 65000, 80000, 42000],
    'YearsAtCompany':  [5, 2, 7, 10, 1, 6, 8, 3],
    'Quit':            [0, 1, 0, 0, 1, 0, 0, 1]
})

print('Original DataFrame:')
print(employees)
print()

# Task 1: Label Encode 'Satisfaction' respecting natural order (Low=0, Medium=1, High=2)
# (Using a manual mapping — sklearn's LabelEncoder would sort alphabetically.)
satisfaction_map = {'Low': 0, 'Medium': 1, 'High': 2}
employees['Satisfaction'] = employees['Satisfaction'].map(satisfaction_map)

print('After Label Encoding Satisfaction:')
print(employees)
print()

# Task 2: Dummy Encode 'Department' (nominal — no ordering)
employees = pd.get_dummies(employees, columns=['Department'], drop_first=True, dtype=int)

print('After Dummy Encoding Department:')
print(employees)
print()

# Task 3: Standard Scale 'Salary' and 'YearsAtCompany'
scaler = StandardScaler()
employees[['Salary', 'YearsAtCompany']] = scaler.fit_transform(employees[['Salary', 'YearsAtCompany']])

# Task 4: Print final result
print('Final Processed DataFrame:')
print(employees.round(3))

# Task 5: What would change if we used Random Forest instead of Logistic Regression?
# Answer:
# - Scaling is unnecessary. Random Forest splits on thresholds and is scale-invariant,
#   so StandardScaler / MinMaxScaler don't help (or hurt) tree-based models.
# - Dummy Encoding's drop_first is optional. Linear models care about multicollinearity
#   because it inflates coefficient variance — trees don't. Plain One-Hot, or even
#   simple Label Encoding for Department, would also be fine for Random Forest.
# - Ordinal encoding for Satisfaction is still a good idea: it gives the tree fewer,
#   more meaningful splits ("Satisfaction >= 2") than one column per category.
```

## 📋 Quick Reference Summary

| Method | Function | New Columns | Use When |
|---|---|---|---|
| **One-Hot Encoding** | `pd.get_dummies()` | N (one per category) | Nominal, low cardinality |
| **Dummy Encoding** | `pd.get_dummies(drop_first=True)` | N-1 | Linear models, avoid multicollinearity |
| **Label Encoding** | `LabelEncoder()` | 0 (in-place) | Ordinal categories only |
| **Standard Scaling** | `StandardScaler()` | 0 (in-place) | Normal distribution, PCA, SVM |
| **Min-Max Scaling** | `MinMaxScaler()` | 0 (in-place) | Neural nets, bounded [0,1] output |

### ✅ Algorithm Cheat Sheet
| Algorithm | Needs Scaling? | Best Encoding |
|---|---|---|
| Linear/Logistic Regression | ✅ Yes | Dummy (drop_first) |
| KNN, SVM | ✅ Yes | One-Hot or Dummy |
| Neural Networks | ✅ Yes (Min-Max) | One-Hot |
| Decision Tree | ❌ No | Label or One-Hot |
| Random Forest, XGBoost | ❌ No | Label or One-Hot |
| PCA | ✅ Yes (Standard) | One-Hot |
